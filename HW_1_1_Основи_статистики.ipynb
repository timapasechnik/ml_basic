**Завдання 3.** Для того ж набору даних в змінній `sum_purchases` обчисліть **медіану** кожним з зазначених методів:  

  3.1. без використання жодних бібліотек, користуючись тільки Python.  
  3.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

```python
#без використання жодних бібліотек, користуючись тільки Python
sorted_sum = sorted(sum_purchases)
n = len(sorted_sum)

if n%2 == 1:
  median = sorted_sum[n//2]
else:
  median = (sorted_sum[(n-1)//2] + sorted_sum[(n)//2])/2

print("Медіана без використання жодних бібліотек =", median)

#використовуючи відповідний метод бібліотеки numpy.
print("Медіана, використовуючи відповідний метод бібліотеки numpy =", np.median(sum_purchases))
```

Вибіркове середнє доцільно використовувати, коли дані не мають значних аномальних значень (викидів).
Медіану доцільно використовувати, коли в даних є дуже великі або дуже малі значення, які можуть спотворювати середнє.

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
#без використання жодних бібліотек, користуючись тільки Python.

#Розраховуємо середні значення та стандартні відхилення
mean_purchases = sum(sum_purchases)/len(sum_purchases)
sigma1 = (sum((i-mean_purchases)**2 for i in sum_purchases)/(len(sum_purchases)-1))**0.5

mean_visit = sum(visits)/len(visits)
sigma2 = (sum((i-mean_visit)**2 for i in visits)/(len(visits)-1))**0.5

#Розраховуємо коваріацію
purch_visits = [(sum_purchases[i] - mean_purchases)*(visits[i] - mean_visit) for i in range(len(sum_purchases))]
covar = sum(purch_visits)/(len(sum_purchases)-1)

#Розраховуємо кореляцію
corr_coef = covar / (sigma1 * sigma2)

#використовуючи відповідний метод бібліотеки numpy.
corr_coef2 = np.corrcoef(sum_purchases,visits)[0,1]

print("Коефіцієнт кореляції Пірсона без використання жодних бібліотек =", corr_coef)
print("Коефіцієнт кореляції Пірсона, використовуючи відповідний метод бібліотеки numpy =", corr_coef2)

```

Спостерігається слабка негативна кореляція: чим частіше покупець відвідує магазин, тим менше він витрачає. Але цього напевно неможна стверджувати, бо зв`язок між показниками слабкий.

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

На графіку розподілу покупок і кількості відвідувань помітні окремі аномально великі значення суми покупок. Вони можуть суттєво впливати на коефіцієнт кореляції Пірсона та спотворювати оцінку лінійної залежності між показниками.

Але видаляти такі спостереження без додаткового обґрунтування не варто, оскільки вони можуть відображати реальну поведінку покупців. Можливо, мало сенс класифікувати такі суми, як окрему групу покупців. То ж все залежить від задачі бізнесу, яку треба вирішити.

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
sum_purchases = np.array(sum_purchases)
visits = np.array(visits)

#Розрахунок threshold
threshold = np.mean(sum_purchases) + 2 * np.std(sum_purchases)

#Прибирання викидів
mask = sum_purchases <= threshold
sum_filtered = sum_purchases[mask]
visits_filtered = visits[mask]

#Коефіцієнт кореляції
new_coefCorr = np.corrcoef(sum_filtered,visits_filtered)[0,1]
print("Коеф. кореляції Пірсона з допомогою numpy для кількості візитів і суми покупок без записів-викидів\n", new_coefCorr)
print("\n")

plt.figure(figsize=(10, 6))
plt.scatter(sum_filtered, visits_filtered)
plt.title('Scatter Plot of Purchases vs. Visits (filtered)')
plt.xlabel('Purchases')
plt.ylabel('Visits')
plt.grid(True)
plt.show()
```

Після видалення викидів із масиву суми покупок коефіцієнт кореляції Пірсона змінився з -0.3027 до -0.3685.
Це свідчить про посилення слабкої негативної лінійної залежності між кількістю візитів і сумою покупок і, що зі збільшенням кількості візитів, сума покупок має тенденцію до зменшення. Проте залежність залишається слабкою.
